```python
# !wget https://raw.githubusercontent.com/karpathy/char-rnn/master/data/tinyshakespeare/input.txt
with open('input.txt', 'r') as f:
    text = f.read()
print(f"length of dataset in characters: {len(text):,}")
```

```python
data = text[:1000]

import tiktoken
enc = tiktoken.get_encoding("gpt2")
tokens = enc.encode(data)
print(tokens[:24])
```

```python
import torch
buf = torch.tensor(tokens[:24]) 
x = buf.view(4, 6)
print(x)
```

```python
import torch
buf = torch.tensor(tokens[:24 + 1]) # last value is for the ground truth of the last token
x = buf[:-1].view(4, 6)
y = buf[1:].view(4, 6)
print(x)
print(y)
```
